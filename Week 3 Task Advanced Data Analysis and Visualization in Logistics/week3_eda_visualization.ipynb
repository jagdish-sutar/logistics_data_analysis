This notebook performs exploratory data analysis and visualization on a **hypothetical logistics dataset** containing 500 shipment records. The analysis focuses on delivery performance, shipment volume, transport cost, delays, regions, transport modes, and customer satisfaction.

**Note:** The dataset is simulated for educational purposes and does not represent a real logistics operation.

## 1. Dataset Structure

Key variables:

- `Region` — geographical operating region
- `Transport_Mode` — Road, Rail, or Air
- `Month` — shipment month
- `Shipment_Volume` — shipment volume indicator
- `Distance_KM` — route distance
- `Delivery_Time_Days` — delivery duration
- `Transport_Cost` — transportation cost indicator
- `Delayed` — binary delay indicator
- `Customer_Satisfaction` — simulated satisfaction score

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('data/hypothetical_logistics_dataset.csv')
df.head()
```

## 2. Basic Data Inspection

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe(include='all'))
```

## 3. Central Tendency and Key Logistics Metrics

```python
metrics = {
    'Average Delivery Time (days)': df['Delivery_Time_Days'].mean(),
    'Median Delivery Time (days)': df['Delivery_Time_Days'].median(),
    'Average Transport Cost': df['Transport_Cost'].mean(),
    'Average Shipment Volume': df['Shipment_Volume'].mean(),
    'Delay Rate (%)': df['Delayed'].mean() * 100,
    'Average Customer Satisfaction': df['Customer_Satisfaction'].mean()
}
pd.Series(metrics).round(2)
```

## 4. Delivery Time Distribution

A histogram shows the distribution and spread of delivery times.

```python
plt.figure(figsize=(9,5))
plt.hist(df['Delivery_Time_Days'], bins=25, edgecolor='black')
plt.title('Distribution of Delivery Time')
plt.xlabel('Delivery Time (Days)')
plt.ylabel('Number of Shipments')
plt.show()
```

## 5. Transport Cost by Mode

A box plot compares the distribution of transportation costs across Road, Rail, and Air.

```python
modes = ['Road', 'Rail', 'Air']
costs = [df.loc[df['Transport_Mode'] == m, 'Transport_Cost'] for m in modes]
plt.figure(figsize=(9,5))
plt.boxplot(costs, labels=modes)
plt.title('Transport Cost by Transport Mode')
plt.xlabel('Transport Mode')
plt.ylabel('Transport Cost')
plt.show()
```

## 6. Shipment Volume vs Delivery Time

A scatter plot explores the relationship between shipment volume and delivery duration.

```python
plt.figure(figsize=(9,5))
plt.scatter(df['Shipment_Volume'], df['Delivery_Time_Days'], alpha=0.55)
plt.title('Shipment Volume vs Delivery Time')
plt.xlabel('Shipment Volume')
plt.ylabel('Delivery Time (Days)')
plt.show()
```

## 7. Delay Rate by Region

```python
delay_rate = df.groupby('Region')['Delayed'].mean().sort_values(ascending=False) * 100
plt.figure(figsize=(9,5))
delay_rate.plot(kind='bar')
plt.title('Delay Rate by Region')
plt.xlabel('Region')
plt.ylabel('Delayed Shipments (%)')
plt.xticks(rotation=0)
plt.show()
delay_rate.round(2)
```

## 8. Correlation Analysis

The correlation matrix provides an initial view of linear relationships among numerical variables. Correlation does not establish causation.

```python
numeric = df.select_dtypes(include=np.number)
corr = numeric.corr()
plt.figure(figsize=(9,7))
plt.imshow(corr, interpolation='nearest', aspect='auto')
plt.colorbar(label='Correlation')
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha='right')
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
corr.round(2)
```

## 9. Monthly Delivery Performance

```python
month_order = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
monthly = df.groupby('Month')['Delivery_Time_Days'].mean().reindex(month_order)
plt.figure(figsize=(10,5))
plt.plot(monthly.index, monthly.values, marker='o')
plt.title('Average Delivery Time by Month')
plt.xlabel('Month')
plt.ylabel('Average Delivery Time (Days)')
plt.show()
```
