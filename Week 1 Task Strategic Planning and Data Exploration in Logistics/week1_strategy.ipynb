## 3. Key Performance Indicators (KPIs)

### KPI 1 — On-Time Delivery Rate
Measures the percentage of orders delivered on or before the expected delivery date.

$$\text{On-Time Rate} = \frac{\text{Orders Delivered On Time}}{\text{Total Delivered Orders}} \times 100$$

### KPI 2 — Average Delivery Lead Time
Measures the average number of days between order approval/purchase and customer delivery.

$$\text{Average Lead Time} = \frac{\sum \text{Delivery Time}}{\text{Number of Delivered Orders}}$$

### KPI 3 — Freight Cost Ratio
Measures freight cost relative to order value.

$$\text{Freight Cost Ratio} = \frac{\text{Freight Value}}{\text{Order Value}} \times 100$$

### Additional KPI — Average Delay
Measures the average number of days by which late deliveries exceed the expected delivery date.

## 5. Proposed Data Science Methods

### Regression
Regression can estimate a continuous logistics metric such as delivery time. Candidate predictors could include freight cost, distance-related variables, region, product characteristics, and operational timestamps.

### Clustering

### Optimization
Optimization can later evaluate resource-allocation or transportation decisions under constraints such as cost, capacity, and service-level requirements.
